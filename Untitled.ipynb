```python
from statsmodels.stats.proportion import proportions_ztest

# данные
success = [
    df[df['group'] == 'control']['converted'].sum(),
    df[df['group'] == 'treatment']['converted'].sum()
]

nobs = [
    df[df['group'] == 'control']['converted'].count(),
    df[df['group'] == 'treatment']['converted'].count()
]

# z-test
z_stat, p_value = proportions_ztest(success, nobs)

z_stat, p_value
```
